### **[], .loc, .iloc**

In this notebook we learn the 3 ways to select data from a DataFrame:

| Method | Selects by | Example |
|--------|-----------|---------|
| `df["col"]` | column name | `df["Age"]` |
| `df.loc[...]` | **labels** (names) | `df.loc[0:4, ["Name","Age"]]` |
| `df.iloc[...]` | **positions** (numbers) | `df.iloc[0:5, 0:3]` |

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\HP\Downloads\titanic_dataset.csv") # change the path if needed
df
```

```python
# Coding
df.head(10)
```

### **Selecting Columns**

### **Single Column** - returns a Series (1 column)

```python
# Coding
df["Age"]
```

### **List of Columns** - returns a DataFrame (note the double brackets!)

```python
# Coding 
df[["Name", "Age", "PassengerId"]]
```

### **.loc** — Label-based Selection (VERY IMPORTANT)

Syntax: **df.loc[row_label, column_label]**

**.loc** uses the **index labels** and column **names**.

### Access Rows with labels: - loc includes the end - 5 rows!

```python
# Coding 
df.loc[0:4]
```

### Access Rows + specific columns together

```python
# Coding 
df.loc[0:4,["Name", "Sex", "Age", "Embarked"]]
```

### Access One selected Rows and selected Column

```python
# Coding
df.loc[4, ["Name", "Age", "Embarked"]]
```

### Access all Rows and specific Columns - the ':' means everything

```python
# Coding
df.loc[:, ["Name", "Pclass", "Survived", "Fare"]]
```

### **.iloc** — Position-based Selection

Syntax: **df.iloc[row_position, column_position]**

**.iloc** uses **integer positions** (0, 1, 2, ...) — like a list.

### Access First 5 Rows (iloc excludes the end - position 0,1,2,3,4)

```python
# Coding
df.iloc[0:5]
```

### Access Rows and Columns by position - first 5 rows and first 3 columns

```python
# Coding 
df.iloc[0:5, 0:3]
```

### **The Key Difference: **loc** vs **iloc**

| | `.loc[0:4]` | `.iloc[0:4]` |
|---|---|---|
| Selects by | label | position |
| End included? | **YES** (5 rows) | **NO** (4 rows) |

**Tip: GENERALLY USE `.loc` INSTEAD OF `.iloc`** — labels do not change when the data is
sorted or filtered, positions do.

### **set_index** — Now Labels and Positions Are DIFFERENT

Until now the index labels (0,1,2,...) happened to equal the positions, so `loc` and `iloc`
looked the same. Watch what happens when we make **PassengerId** the index.

```python
# Coding
df.set_index("PassengerId", inplace = True)
df.head()
```

### Now the index labels are 1 to 891 (Passenger Id), but positions are still 0 to 890


```python
# Coding
df.loc[5]     # LABEL 1  -> passenger with PassengerId = 1  (the FIRST row)
```

```python
# Coding
df.iloc[4]      # POSITION 887 -> the 888th row (its PassengerId label is 888)
# note* position index and label index are different in number. 
```

### Same Passenger, two ways:

```python
# Coding
print("By label   ->  df.loc[888]")
print("By position ->  df.iloc[887]")
df.loc[888].equals(df.iloc[887])
```

### **Summary**

- `df["col"]` -> one column (Series); `df[["c1","c2"]]` -> many columns (DataFrame)
- `.loc[rows, cols]` -> by **labels**, end **included**
- `.iloc[rows, cols]` -> by **positions**, end **excluded**
- After `set_index`, labels and positions become different — `.loc` follows the new index
- Prefer `.loc` in daily work
