## **Pandas_Groupby Notebook**

### **What is GroupBy?**

A groupby() is used to split data into groups, apply a function, and combine results.

Split → Apply → Combine

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Department": ['IT', 'HR', "IT", "Finance", "HR", "Finance", "IT"], 
    "Employee": ["A", 'B', 'C', 'D', 'E', 'F', 'G'], 
    'Salary': [50000, 40000, 55000, 60000, 42000, 65000, 58000],
    'Experience': [2, 1, 3, 5, 2, 6, 4]
})

df 
```

### **Simple Groupby** 

```python
# Coding
df.groupby('Department')
```

### Calculate mean salary per Department using groupby()

```python
# Coding 
df.groupby('Department')["Salary"].mean()
```

### The multiple aggregation using .groupby(); group **'Department'** and find aggregation of **'Salary'**

```python
# Coding 
df.groupby("Department")["Salary"].agg(["mean", "max", "min", "sum"])
```

### Grouping multiple columns using .groupby(); Column **'Department** and **'Experience'** and find mean Salary

```python
# Coding
df.groupby(["Department", "Experience"])["Salary"].mean()
```

### Count the numbers of employee in different department by using .groupby()

```python
# Coding 
df.groupby("Department")["Employee"]. count()
```

### Finding a **salary_range** using .groupby() also calculating max and min salary to find the range. 

```python
def salary_range(x):
    return x.max()- x.min()

df.groupby("Department")["Salary"].apply(salary_range)
```
