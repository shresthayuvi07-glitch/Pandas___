## **Pandas_Index_Set and Reset Notebook**

````python
# Coding
import pandas as pd
df = pd.DataFrame({
    "Name": ["James", "Lilly", "Harry", "Peter", "Jack", "Ron", "Adin"], 
    "Age": [20, 24, 25, 26, 22, 21, 28], 
    "City": ["Paris", "Vienna", "Berlin", "Helsinki", "Tokyo", "London", "Oslo"], 
    "Occuption": ["Plumber", "IT", "Sales Person", "Manager", "Driver", "Designer", "Doctor"]
})
df
````

### **Default Index**
By Default index, pandas always assigns a numeric index starting from 0


````python
# Coding to check the index:
print(df.index)
````

### **Set a Column as an index**


````python
# Coding
df.set_index("Name")
````

### **To set index permanently** 

````python
# Coding 
df.set_index("Name", inplace=True)
df
````

### **To Reset Index**

````python
# Coding
df.reset_index() 
# This is only temporary reset index.
````

````python
df
````

### **To Reset index permanently** 

````python
# Coding
df.reset_index(inplace = True)
df
````

### **What does inplace=True mean?**

- **inplace=True** means **modify the original DataFrame**
- No new DataFrame is created
- Changes happen **permanently**
