### **Pandas loc and iloc**:

````python
# Example:
import pandas as pd
data = {
    "Name": ["James", "Mann", "Tony", "Peter"], 
    "Id": [101, 102, 103, 104], 
    "Marks": [50, 60, 70, 85]
}
df = pd.DataFrame(data)
df
````

### **loc - Label Base Indexing**:

- **loc** is used to access data using **labels**
- Labels can be:
  - Index names /row label
  - Column names
- **Both start and end values are included**


````python
# Example; 
df.set_index("Id", inplace=True)
df
````

**loc = access data using label index name value, not index position**

### Accessing Single or specific rows using loc

````python
# Example: 
df.loc[103,"Name"]
````

### Accessing Double and multiple rows using loc

````python
# Exmaple:
df.loc[[103, 104]]   
# Access multiple rows
````

### Accessing Multiple rows and column using loc

````python
# Example:
df.loc[101:103,["Name", "Marks"]]
````

### **iloc - Integer Position based Indexing**:

- **iloc** uses **integer positions**
- Index starts from **0**
- **End value is excluded**
- Similar to Python list slicing


````python
# Example:
import pandas as pd
data = {
    "Name": ["James", "Lilly", "Thor", "Peter"], 
    "Id": [101, 102, 103, 104], 
    "Marks": [50, 60, 70, 85]
}
df = pd.DataFrame(data)
df
````

````python
# Example:
df.set_index("Marks") 
df
````

### Access multiple rows using iloc

````python
# Exmaple: 
df.iloc[[0,2]]
````

### Access specific rows and column using iloc

````python
# Example: 
df.iloc[2,1]
````

### Access Range of Rows and Column using iloc

````python
# Example: 
df.iloc[0:3, 0:2]
````

### **Comparison Table**

| Feature | loc | iloc |
|------|-----|------|
| Type | Label-based | Integer position-based |
| Uses | Index & column names | Row & column numbers |
| Range | End included | End excluded |
| Similar to | Dictionary | Python list |
| Example | df.loc[101] | df.iloc[0] |

### When to Use **loc**
- When index has meaningful labels (ID, Name, Date)
- When readability is important

### When to Use **iloc**
