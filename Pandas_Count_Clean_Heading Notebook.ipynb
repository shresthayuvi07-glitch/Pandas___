### **Import Pandas and Create Sample Data**

We will first use a small student dataset so the counting methods would be easy to understand.


```python
import pandas as pd 
df = pd.DataFrame({
    
    "Name": ["Anita", "Bibek", "Chandra", "Dipesh", "Elina", "Farhan"],
    "Department": ["CSIT", "BCA", "CSIT", "BIM", "BCA", "CSIT"],
    "Age": [22, 21, 23, 24, 22, 25],
    "Score": [85, 72, 91, None, 68, 77],
    "City": ["Kathmandu", "Pokhara", "Kathmandu", "Butwal", "Pokhara", None]
})
df
```

### **Count Total Rows with len()**

**len(df)** counts the total number of rows. It counts a row even if some values in that row are missing.


```python
# Coding 
len(df)
```

### Count the total number of **Students** in the DataFrame

```python
len(df["Name"])
```

### **Count Non-Missing Values with .count()**

**count()** counts only values that are not missing. If a column has **NaN**, those missing values are not counted.


```python
# Coding 
df["Score"].count()
```

### Count the non-missing values in the cloumn - **"City"**

```python
# Coding 
df["City"].count()
```

### **Count Non-Missing Values in Every Column**

**df.count()** returns the number of non-missing values for each column.


```python
# Coding 
df.count()
```

### **Count Values Row by Row**

Use **axis=1** to count non-missing values across each row.


```python
# Coding 
df.count(axis=1)
```

### **Count Frequency with value_counts()**

**value_counts()** tells us how many times each unique value appears.


```python
# Coding 
df["Department"].value_counts()
```

### Count how many **Students** come from each **City**

```python
# Coding 
df["City"].value_counts()
```

```python
# Coding 
df.groupby("City")["Name"].count()
```

### **Include Missing Values in value_counts()**

By default, **value_counts()** ignores missing values.

Use **dropna=False** to include them.


### Count **City** value including missing values

```python
# Coding 
df["City"].value_counts(dropna= False)
```

### Count **Department** values and include missing values if any.

```python
# Coding 
df["Department"].value_counts(dropna = False)
```

### **Count Unique Values with .nunique()**

**.nunique()** tells us how many different values exist.


### Count unique values from **Department**

```python
# Coding 
df["Department"].nunique()
```

### Count how many unique **City** values are present.

```python
# Coding 
df["City"].nunique()
```

### **Count Rows with a Condition**

A condition returns **True** or **False**.

In Pandas:

- **True** behaves like **1**
- **False** behaves like **0**

So **.sum()** can count how many rows satisfy the condition.


### Count **Students** whose Score is 80 or more

```python
# Coding 
(df["Score"]>=80).sum()

# Note* .sum() will always count the value which are true, but .count() will always count the value which are both true and false. 
```

```python
# Coding 
df.loc[(df["Score"]> 80),:]
```

### Count how many **Students** are from the **CSIT Department**.

```python
# Coding 
(df["Department"]== "CSIT").sum()
```

```python
# Coding 
df.loc[(df["Department"]== "CSIT"), :]
```

### **Count with Multiple Conditions**

Use **&** when both conditions must be true.


### Count **CSIT students** whose Score is **80** or more

```python
((df["Department"] == "CSIT") & (df["Score"]>= 80)).sum()
```

```python
# Coding 
df.loc[(df["Department"]== "CSIT") & (df["Score"]>= 80), :]
```

### Count **Students** who are from **Pokhara** and have **Age greater than 20**.

```python
# Coding 
((df["City"]== "Pokhara") & (df["Age"]> 20)).sum()
```

```python
# Coding 
df.loc[(df["City"]== "Pokhara") & (df["Age"]> 20), :]
```

### **Count Groups with .groupby().count()**

**groupby().count()** counts non-missing values inside each group.


### Count available **Score** values inside each **Department**

```python
# Coding 
df.groupby("Department")["Score"].count()
```

### Count available **City** values inside each **Department**.

```python
# Coding 
df.groupby("Department")['City'].count()
```

### **Count Total Rows per Group with .groupby().size()**

**size()** counts every row inside each group. This is different from **count()**, because **count()** ignores missing values.


### Count total students in each **Department**

```python
# Coding 
df.groupby("Department")["Name"].size()
```

### Count total rows for each **City** using groupby().size().

```python
# Coding 
df.groupby("City").size()
```

### **Counting Patterns to Remember**

**len(df)**  
→ total rows

**df["column"].count()**  
→ non-missing values

**df["column"].value_counts()**  
→ frequency of each value

**df["column"].nunique()**  
→ number of unique values

**(condition).sum()** 
→ count rows matching a condition

**df.groupby("column").size()**  
→ total rows in each group

**df.groupby("column")["value"].count()**  
→ non-missing values inside each group

**df.isna().sum()**  
→ count missing values

**pd.crosstab(col1, col2)**  
→ frequency table
