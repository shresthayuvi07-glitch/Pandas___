### **Syntax for loc** 
**df.loc[row_condition, column_selection]**

### Accessing Age above 30:

```python
# Coding
df[df["Age"]>30]
```

### Using .loc

```python
# Coding 
df.loc[df["Age"]>30, :]
```

### Access data to find a Passengers who has survived

```python
# Coding 
df.loc[df["Survived"]==1, ["Name", "Age", "Survived"]]
```

### Access data to find Passenger who are in first Class

```python
# Coding 
df.loc[df["Pclass"]==1, ["Name", "Age", "Sex", "Pclass"]]
```

### Accessing data using **and condition**:

### Q.1 Find out a Passenger data who are above age of 30 and has survived

```python
# Caoding 
df.loc[(df["Age"]>30) & (df["Survived"]==1), :]
```

### Q.2 Find out a Passenger data whose age is above 18 or must be Female Passenger 

```python
# Coding
df.loc[(df["Age"]>25) | (df["Sex"]=="female"), :]
```

### Q.3 Find out a data of Passenger who are in 3-class and is female

```python
# Coding
df.loc[(df["Pclass"]==3) & (df["Sex"]== "female"), ["Name", "Age", "Pclass"]]
```

### Q.4 Find out a data of Passenger who are in 3-class and is male

```python
# Coding 
df.loc[(df["Pclass"]==3) & (df["Sex"]=="male"), ["Name", "Age", "Pclass"]]
```

### Q.5 Find out the data of female Passenger who are olde than 25

```python
# Coding 
df.loc[(df["Age"]>25) & (df["Sex"]== "female"), ["Name", "Age", "Sex"]]
```

### Q.6 Find out the average age of Survival

```python
# Coding
df.loc[df["Survived"]==1, "Age"].mean()
```

### Q.7 Find out the data of Rich passengers (Fare > 100) who survived, only find name and age


```python
# Coding
df.loc[(df["Fare"]>100) & (df["Survived"]==1), ["Name", "Age", "Fare"]]
```

### Q.8 Find out the data of the name and age of Children (<15) in 3rd class

```python
# Coding
df.loc[(df["Age"]<15) & (df["Pclass"]==3), ["Name", "Age"]]
```

### Q.9 Count passengers older than 60

```python
# Coding
df.loc[df["Age"]>60, :].shape[0]
```
