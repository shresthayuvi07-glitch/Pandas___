## **Pandas_Loc and Iloc Notebook**

### **Detail Comparision with Examples**:

````python
# Coding 
import pandas as pd
df = pd.DataFrame({
    "Name": ["Mann", "James", "Jude", "lilly", "Lana", "Parker"], 
    "ID": [101, 102, 103, 104, 105, 106], 
    "Marks": [78, 80, 90, 67, 69, 70]
})
df
````

### **loc** – Label-based Indexing

- **loc** is used to access data using **labels**
- Labels can be:
  - Index names /row label
  - Column names
- **Both start and end values are included**


````python
# Coding 
df.set_index("ID", inplace = True)
df
````

### **loc=access data using label index name value not index postion**

### Accessing value from rows and columns

````python
# Coding
df.loc[103, ["Name", "Marks"]]
````

### Accessing multiple rows value

````python
# Coding 
df.loc[[101, 105, 106]]
````

### Accessing values from specific rows

````python
#Coding 
df.loc[103]
````

### Accessing values using range

````python
# Coding 
df.loc[102:105, ["Name", "Marks"]]
````

## **iloc** – Integer Position-based Indexing

- **iloc** uses **integer positions**
- Index starts from **0**
- **End value is excluded**
- Similar to Python list slicing


````python
# Coding
df.set_index("Marks")
df
````

### Accessing multiple rows using iloc

````python
# Coding
df.iloc[[0,2, 5]]
````

### Accessing specific rows and column 


````python
# Coding
df.iloc[3,0]
````

### Accessing range of rows and column

````python
# Coding
df.iloc[0:3, 0:2]
````

## **Comparison Table**

| Feature | loc | iloc |
|------|-----|------|
| Type | Label-based | Integer position-based |
| Uses | Index & column names | Row & column numbers |
| Range | End included | End excluded |
| Similar to | Dictionary | Python list |
| Example | df.loc[101] | df.iloc[0] |

### When to Use **loc**
- When index has meaningful labels (ID, Name, Date)
- When readability is important

### When to Use **iloc**
