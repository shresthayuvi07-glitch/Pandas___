## **Pandas_Sorting Notebook**

```python
# Coding
import pandas as pd
df = pd.DataFrame({
    "Name": ["James", "Lilly", "Harry", "Peter", "Jack", "Ron", "Adin"], 
    "Age": [20, 24, 25, 26, 22, 21, 28], 
    "City": ["Paris", "Vienna", "Berlin", "Helsinki", "Tokyo", "London", "Oslo"], 
    "Marks": [68, 90, 80, 75, 78, 85, 60]
})
df
```

### Use inplace = True to change data permannently

```python
# Coding 
df.set_index("Name", inplace = True)
df
```

### Sort the data by Age in ascending order

```python
# Coding 
df.sort_values(by= "Age")
```

### Sort the data by Age and Marks in Ascending order, but if the age is repeated, what will be the output?

```python
# Coding
import pandas as pd
df = pd.DataFrame({
    "Name": ["James", "Lilly", "Harry", "Peter", "Jack", "Ron", "Adin"], 
    "Age": [20, 20, 25, 26, 20, 21, 28], 
    "City": ["Paris", "Vienna", "Berlin", "Helsinki", "Tokyo", "London", "Oslo"], 
    "Marks": [68, 90, 80, 75, 78, 85, 60]
})
df
```

### Sort the data by multiple values such as Age and Marks 

```python
df.sort_values(by=["Age", "Marks"])
```

### Here, pandas sort the data from Age not of the Marks, because pandas look for first in, first out.

### In case, the age contain same multiple data, pandas will sort by Marks in ascending order, but still pandas keep the Age in ascending order.

### Sort the data by Marks in descending order

```python
# Coding 
df.sort_values(by="Marks", ascending = False)
```

### **Sorting by Index or, Sort Index**

### Sorting by index in ascending order

```python
# Coding 
df.sort_index()
```

### Sorting by index in descending order

```python
# Coding 
df.sort_index(ascending=False)
```

### **Sorting data by Alphabetical order**

### Sorting by Names

```python
# Coding 
df.sort_values(by= "Name")
```

### Sorting by City

```python
# Coding
df.sort_values("City")
```

## 🧠 Quick Summary

| Task | Method |
|------|--------|
| Sort rows | **sort_values()** |
| Sort index | **sort_index()** |
| Descending order | **ascending=False** |
| Multiple columns | **by=[...]** |
| Save changes | **inplace=True** |
| Fix index | **reset_index(drop=True)** |
